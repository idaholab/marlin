```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def gamma(g):
    g2 = g**2
    gamma_inv = -3.0944 * g2**4 -1.8169*g2**3 + 10.323 * g2**2 - 8.1819*g2 + 2.0033
    return 1 / gamma_inv
```

```python
g = np.linspace(0.098,0.766,1000)

gamma = 1/gamma_inv
plt.plot(gamma,g)

print(min(gamma),max(gamma))
```
